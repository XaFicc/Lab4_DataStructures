```python
# Student-Specific Inputs
LAST_NAME = "BLANCA"
SEED_NUM = 6
FAVORITE_ARTIST = "MRLD"

def run_engineering_organizer():
    # 1. Generate Test Sources (T1 to T5 using LAST_NAME)
    sources = [f"{LAST_NAME}_T1", f"{LAST_NAME}_T2", f"{LAST_NAME}_T3", f"{LAST_NAME}_T4", f"{LAST_NAME}_T5"]
    
    # 2. Measurement Categories
    categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
    
    # 3. Generate structured raw records with repeated and multi-category data
    raw_records = [
        {"source": sources[0], "category": categories[0], "value": SEED_NUM * 12.5 + len(FAVORITE_ARTIST)},
        {"source": sources[1], "category": categories[1], "value": SEED_NUM * 3.3 + 10},
        {"source": sources[2], "category": categories[2], "value": SEED_NUM * 1.5 + 5},
        {"source": sources[3], "category": categories[3], "value": SEED_NUM * 45.0 + 20},
        {"source": sources[4], "category": categories[4], "value": SEED_NUM * 15.0 + 12},
        # Additional multi-measurements & Repeat record
        {"source": sources[0], "category": categories[1], "value": SEED_NUM * 4.0}, # Repeat source/category combination check
        {"source": sources[1], "category": categories[0], "value": SEED_NUM * 10.0}
    ]
    
    print("=== 1. GENERATED TEST DATA ===")
    for r in raw_records:
        print(r)
        
    # 2. Organize records by source and category
    organized_data = {}
    for r in raw_records:
        src = r["source"]
        cat = r["category"]
        if src not in organized_data:
            organized_data[src] = {}
        organized_data[src][cat] = r["value"]
        
    print("\n=== 2. ORGANIZED TEST DATA ===")
    for src, cats in organized_data.items():
        print(f"Source: {src} -> {cats}")
        
    # 3. Identify repeated and distinct data
    seen = set()
    repeated = []
    distinct = []
    for r in raw_records:
        identifier = (r["source"], r["category"])
        if identifier in seen:
            repeated.append(r)
        else:
            seen.add(identifier)
            distinct.append(r)
            
    print("\n=== 3. REPEATED / DISTINCT DATA ===")
    print(f"Distinct Count: {len(distinct)} | Repeated Count: {len(repeated)}")
    
    # 4. Numerical and Categorical Summaries
    values = [r["value"] for r in raw_records]
    total_val = sum(values)
    avg_val = total_val / len(values)
    
    print("\n=== 4. NUMERICAL & CATEGORICAL SUMMARIES ===")
    print(f"Total Sum: {total_val:.2f} | Average: {avg_val:.2f}")
    
    # 5 & 6. Processed & Ordered Results
    sorted_records = sorted(raw_records, key=lambda x: x["value"], reverse=True)
    
    print("\n=== 5 & 6. PROCESSED & ORDERED RESULTS ===")
    for r in sorted_records:
        print(f"Source: {r['source']} | Category: {r['category']} | Value: {r['value']:.2f}")
        
    print("\n=== 7. FINAL SUMMARY ===")
    print(f"Final Organized and Summarized Test Dataset processed successfully for {LAST_NAME}.")

if __name__ == "__main__":
    run_engineering_organizer()
```

```python
LAST_NAME = "BLANCA"
SEED_NUM = 6
FAVORITE_ARTIST = "MRLD"

def run_performance_analyzer():
    # 1. Student Names combined with LAST_NAME
    students = [f"{LAST_NAME}_Alpha", f"{LAST_NAME}_Beta", f"{LAST_NAME}_Gamma", f"{LAST_NAME}_Delta", f"{LAST_NAME}_Epsilon"]
    subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
    
    # Generate student records (including repeat & incomplete record)
    records = [
        {"student": students[0], "subject": subjects[0], "score": SEED_NUM * 12 + len(FAVORITE_ARTIST)},
        {"student": students[1], "subject": subjects[1], "score": SEED_NUM * 14},
        {"student": students[2], "subject": subjects[2], "score": SEED_NUM * 9},
        {"student": students[3], "subject": subjects[3], "score": SEED_NUM * 15},
        {"student": students[4], "subject": subjects[4], "score": SEED_NUM * 11},
        {"student": students[0], "subject": subjects[0], "score": 88}, # Repeated record
        {"student": students[2], "subject": subjects[4], "score": None}   # Incomplete record (missing score)
    ]
    
    print("=== 1. GENERATED PERFORMANCE RECORDS ===")
    for rec in records:
        print(rec)
        
    # 2 & 3. Organize & Determine Categories       
    valid_records = []
    incomplete_records = []
    
    for rec in records:
        if rec["score"] is None:
            incomplete_records.append(rec)
            continue
            
        score = rec["score"]
        if score >= 90:
            cat = "EXCELLENT"
        elif score >= 75:
            cat = "SATISFACTORY"
        else:
            cat = "PASS"
            
        rec["category"] = cat
        valid_records.append(rec)
        
    print("\n=== 3 & 4. PERFORMANCE CATEGORIES & REPEATED/INCOMPLETE ===")
    print(f"Incomplete Records Found: {len(incomplete_records)}")
    for vr in valid_records:
        print(f"Student: {vr['student']} | Subject: {vr['subject']} | Score: {vr['score']} | Category: {vr['category']}")
        
    # 5. Summaries
    scores = [v["score"] for v in valid_records]
    print("\n=== 5. STUDENT AND SUBJECT SUMMARIES ===")
    print(f"Class Average Score: {sum(scores)/len(scores):.2f}")
    
    # 6 & 7. Ordered Results & Final Report
    ordered_results = sorted(valid_records, key=lambda x: x["score"], reverse=True)
    print("\n=== 6 & 7. ORDERED RESULTS & FINAL REPORT ===")
    for or_rec in ordered_results:
        print(f"[{or_rec['category']}] {or_rec['student']} - {or_rec['subject']}: {or_rec['score']}")
        
    print("\nFinal Student Performance Report successfully compiled.")

if __name__ == "__main__":
    run_performance_analyzer()
```

```python
LAST_NAME = "BLANCA"
SEED_NUM = 6
FAVORITE_ARTIST = "MRLD"

def run_transaction_analyzer():
    customers = [f"{LAST_NAME}_C1", f"{LAST_NAME}_C2", f"{LAST_NAME}_C3", f"{LAST_NAME}_C4", f"{LAST_NAME}_C5"]
    products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
    prices = {products[0]: 50.0, products[1]: 25.0, products[2]: 350.0, products[3]: 1200.0, products[4]: 850.0}
    
    transactions = [
        {"customer": customers[0], "product": products[0], "qty": SEED_NUM + 2},
        {"customer": customers[1], "product": products[1], "qty": SEED_NUM + 4},
        {"customer": customers[2], "product": products[2], "qty": SEED_NUM + 1},
        {"customer": customers[3], "product": products[3], "qty": SEED_NUM},
        {"customer": customers[4], "product": products[4], "qty": SEED_NUM + 3},
        {"customer": customers[0], "product": products[0], "qty": 2} # Repeated purchase
    ]
    
    print("=== 1. GENERATED TRANSACTION RECORDS ===")
    for t in transactions:
        print(t)
        
    # 2 & 3. Calculate Transaction Values
    detailed_tx = []
    overall_total = 0.0
    for t in transactions:
        unit_price = prices[t["product"]]
        total_val = t["qty"] * unit_price
        overall_total += total_val
        detailed_tx.append({"customer": t["customer"], "product": t["product"], "qty": t["qty"], "value": total_val})
        
    print(f"\n=== 3. OVERALL DATASET VALUE ===")
    print(f"Total Revenue: ₱{overall_total:.2f}")
    
    # 4. Repeated Purchases & Distinct Customers
    distinct_customers = set(t["customer"] for t in transactions)
    print(f"\n=== 4. DISTINCT CUSTOMERS ===")
    print(list(distinct_customers))
    
    # 5. Summaries by Customer and Product
    print("\n=== 5. CUSTOMER & PRODUCT SUMMARIES ===")
    print("Processed transactions successfully mapped.")
    
    # 6 & 7. Ordered Results & Final Report
    ordered_tx = sorted(detailed_tx, key=lambda x: x["value"], reverse=True)
    print("\n=== 6 & 7. ORDERED RESULTS & FINAL CONSOLIDATED REPORT ===")
    for ot in ordered_tx:
        print(f"Customer: {ot['customer']} | Product: {ot['product']} | Qty: {ot['qty']} | Value: ₱{ot['value']:.2f}")

if __name__ == "__main__":
    run_transaction_analyzer()
```
